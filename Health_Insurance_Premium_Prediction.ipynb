# Health Insurance Premium Prediction

Beginner notebook for EDA and understanding the data. The included dataset is synthetic and for learning only.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

ROOT = Path.cwd()
if not (ROOT / 'data' / 'insurance.csv').exists():
    ROOT = ROOT.parent
df = pd.read_csv(ROOT / 'data' / 'insurance.csv')
df.head()
```

## Inspect the dataset

```python
print('Shape:', df.shape)
df.info()
df.describe(include='all')
```

## Missing values and duplicates

```python
print(df.isnull().sum())
print('Duplicate rows:', df.duplicated().sum())
```

## Target distribution

```python
plt.figure(figsize=(8, 4))
sns.histplot(df['charges'], kde=True)
plt.title('Distribution of synthetic insurance charges')
plt.xlabel('Charges')
plt.tight_layout()
plt.show()
```

## Compare charges by smoking status

```python
plt.figure(figsize=(7, 4))
sns.boxplot(data=df, x='smoker', y='charges')
plt.title('Synthetic charges by smoking status')
plt.tight_layout()
plt.show()
```

## Correlation between numeric features

```python
plt.figure(figsize=(7, 4))
sns.heatmap(df.select_dtypes(include='number').corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Numeric feature correlation')
plt.tight_layout()
plt.show()
```
